**Objetivo:** Analizar las propiedades físicas de exoplanetas confirmados detectados por TESS y visualizar estadísticas básicas como radio, masa y periodo orbital.  

Se eligió este dataset por ser **abierto y confiable**, con información actualizada sobre exoplanetas detectados por la misión TESS, lo cual permite practicar análisis de datos astronómicos reales.


```python
import pandas as pd

# Configurar pandas para mostrar todas las columnas
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)

# Cargar CSV, saltando las líneas que empiezan con #
df = pd.read_csv("PS_2025.09.04_15.01.23.csv", comment='#')

# Vista previa de los datos
print("📄 Primeras filas del dataset:")
print(df.head())

# Información general del dataset
print("\nℹ️ Información del dataset:")
print(df.info())

# Estadísticas básicas
print("\n📊 Estadísticas básicas:")
print(df.describe())

```

```python
import matplotlib.pyplot as plt

# Conteo de planetas por año de descubrimiento
planetas_por_ano = df['disc_year'].value_counts().sort_index()

# Gráfico
plt.figure(figsize=(10,5))
planetas_por_ano.plot(kind='bar', color='skyblue')
plt.title("Número de exoplanetas descubiertos por año")
plt.xlabel("Año")
plt.ylabel("Cantidad de planetas")
plt.show()
```

```python
plt.figure(figsize=(10,5))
df['pl_rade'].hist(bins=30, color='salmon')
plt.title("Distribución de radios de los planetas (en radios terrestres)")
plt.xlabel("Radio del planeta")
plt.ylabel("Cantidad de planetas")
plt.show()
```

```python
metodos = df['discoverymethod'].value_counts()

plt.figure(figsize=(10,5))
metodos.plot(kind='bar', color='lightgreen')
plt.title("Cantidad de planetas por método de descubrimiento")
plt.xlabel("Método")
plt.ylabel("Cantidad de planetas")
plt.show()
```
