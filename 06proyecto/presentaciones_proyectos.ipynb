## 3. Proyectos Presentados

### Grupo 1 – Lengua de Señas (Paula, Hugo, Franco, Ariel, Eduardo y Nicolás)
- **Objetivo:** crear la primera app educativa para aprender **lengua de señas chilena**, iniciando con el alfabeto.  
- **Tecnologías:** YOLOv8, MediaPipe, Django, datasets en RoboFlow.  
- **Resultados:** prototipo web funcional con dos modos (detección libre y práctica gamificada).  
- **Desafíos:** falta de datasets estandarizados y variaciones regionales.  
- **Proyecciones:** ampliar a palabras y frases, integración con plataformas educativas, despliegue móvil.  
- **Comentarios:** invitados destacaron el valor inclusivo y la posibilidad de escalar hacia un traductor en tiempo real.  

---

### Proyecto de Felipe – Detección de Atención en el Aula
- **Objetivo:** medir niveles de atención (alta, media, baja) en clases mediante visión computacional.  
- **Metodología:** YOLOv8 + keypoints; sistema de puntuación de posturas (recto, encorvado, uso de celular).  
- **Resultados:** demo en video con métricas de atención.  
- **Desafíos:** diferenciar entre usos legítimos del celular y distracciones.  
- **Proyecciones:** reportes automáticos para docentes, integración con cámaras existentes.  
- **Comentarios:** se valoró la innovación pero se advirtió sobre implicaciones éticas y de privacidad.  

---

### Proyecto de Sebastián – CADIS-AI (Diagnóstico de Cáncer)
- **Objetivo:** usar CNN para apoyar el diagnóstico temprano de cáncer de mama en imágenes histopatológicas.  
- **Tecnologías:** TensorFlow/Keras, Kaggle, Streamlit.  
- **Resultados:** accuracy de hasta 86%; app con selector de modelos en Streamlit.  
- **Desafíos:** precisión inferior a la mamografía digital estándar (~93%).  
- **Proyecciones:** optimizar hiperparámetros, incluir métricas de falsos positivos/negativos, expandir a otros tipos de cáncer.  
- **Comentarios:** recomendado enfatizar el rol de apoyo clínico y explorar concursos de innovación en salud.  

---

### Proyecto de Alicia – Sketch-KET (Predicción de Trazos)
- **Objetivo:** asistente para artistas que sugiera trazos en tiempo real.  
- **Tecnologías:** Transformers, QuickDraw dataset, Python, Canvas HTML5.  
- **Resultados:** dos modelos entrenados (básico y avanzado), demo funcional con sugerencias en rojo, accuracy ~85%.  
- **Desafíos:** entrenamientos largos en Colab y predicciones aún limitadas.  
- **Proyecciones:** más datos y épocas, incorporación de estilos artísticos, app motivacional.  
- **Comentarios:** se valoró la innovación pedagógica y el potencial creativo.  

---

### Proyecto de Fernando – Detección de Patentes Vehiculares
- **Objetivo:** entrenar un modelo para leer y verificar patentes en tránsito.  
- **Tecnologías:** YOLO + OCR, dataset propio con miles de imágenes etiquetadas.  
- **Resultados:** demo de detección en video, con reconocimiento de caracteres y alertas.  
- **Desafíos:** etiquetas incompletas y rendimiento en condiciones reales.  
- **Proyecciones:** integración en soluciones de seguridad vial y transporte.  

---

### Proyecto de Kiara – Detección de Humo y Fuego
- **Objetivo:** construir un modelo de visión computacional para diferenciar humo de fuego.  
- **Tecnologías:** datasets de RoboFlow, entrenamiento en Colab.  
- **Resultados:** prototipo funcional en imágenes y video.  
- **Desafíos:** falsos positivos y limitaciones de hardware en entrenamiento.  
- **Proyecciones:** despliegue en Vercel y validación con datasets más grandes.  

---

### Proyecto de Leandro – Recomendador de Empleos
- **Objetivo:** crear un sistema de recomendación de empleos a partir de scraping de portales.  
- **Metodología:** scraping, vectorización y cosine similarity para encontrar empleos afines al perfil.  
- **Resultados:** prototipo que recomienda 10 empleos relevantes.  
- **Desafíos:** algunos enlaces no llevan al detalle de la oferta.  
- **Proyecciones:** ampliar scraping a más portales, app en Gradio.  

---

### Proyecto Ecosort – Clasificación de Residuos
- **Objetivo:** prototipo educativo para clasificar basura en colegios.  
- **Tecnologías:** dataset RecycleTrash de RoboFlow, YOLOv8s.  
- **Resultados:** demo inicial con 5 categorías (plástico, cartón, papel, vidrio, metal).  
- **Desafíos:** dataset pequeño, necesidad de entrenamiento más robusto.  
- **Proyecciones:** escalar como solución de robótica educativa.
