## 2. Proyecto de Hugo, Franco e Iván – App de Lengua de Señas
- **Avances:**
  - Intentaron deployar la app Django en **Railway**, **Vercel** y **Render**.
  - Lograron conectar repositorios desde GitHub.
  - Evaluaron también **PythonAnywhere** como alternativa sencilla.
- **Problemas:**
  - Errores de configuración (`requirements.txt`, versiones de Python).
  - Fallos en Render por dependencias faltantes (ej. Gunicorn).
  - Dificultades con variables de entorno y comandos de ejecución.
- **Recomendaciones:**
  - Crear `runtime.txt` para fijar versión de Python.
  - Generar `requirements.txt` actualizado con todas las dependencias.
  - Probar primero en PythonAnywhere antes de pasar a una opción más compleja.
- **Próximos pasos:**
  - Resolver compatibilidades.
  - Confirmar plataforma definitiva para deploy.
  - Presentar un prototipo accesible.

## 3. Proyecto de Sebastián – Diagnóstico de Cáncer
- **Avances:**
  - Entrenó con checkpoints, evitando repetir sesiones largas.
  - Logró accuracy cercano a **0.83**.

- **Problemas:**
  - Dificultades al cargar el modelo en **Streamlit**.

- **Recomendaciones:**
