## 1. Proyecto de Sebastián – Diagnóstico de Cáncer
- **Objetivo:** entrenar un modelo para diagnóstico temprano de cáncer a partir de imágenes histopatológicas.
- **Dataset:** *Breast Histopathology Images* (más de 270 mil imágenes de 50x50 píxeles).
- **Avances:**
  - Logró cargar y visualizar imágenes.
  - Intentó normalizar las imágenes, pero Colab se quedó sin memoria.
- **Problemas:**
  - Limitaciones técnicas al preprocesar tantas imágenes en Google Colab.
- **Recomendaciones:**
  - Usar **RoboFlow** para gestión y preprocesamiento de datasets grandes.
  - Considerar entrenar con subconjuntos antes de escalar.
- **Próximos pasos:** estabilizar pipeline y avanzar hacia entrenamiento del modelo.

## 2. Proyecto de Kiara – Detección de Humo y Fuego
- **Objetivo:** construir un modelo de visión computacional para diferenciar humo y fuego.
- **Dataset:** imágenes anotadas con humo, fuego y combinaciones.
- **Avances:**
  - Encontró datasets pre-anotados y comenzó a probar modelos.
  - Intentó entrenar en Colab, pero las sesiones eran demasiado cortas.
- **Problemas:**
  - Sesiones de entrenamiento interrumpidas (máximo 12 horas).
  - Procesamiento de imágenes muy pesado en local/Colab.
- **Recomendaciones:**
  - Migrar el flujo de trabajo a **RoboFlow**, especializado en computer vision.
  - Probar modelos ya disponibles como baseline y luego optimizar.
- **Próximos pasos:** usar RoboFlow para manejar dataset, entrenar y evaluar modelos con métricas de precisión/recall.

## 3. Proyecto de Alicia – Predicción de Trazos en Dibujos
- **Objetivo:** entrenar un modelo (inicialmente con RNN) para predecir el siguiente trazo en dibujos tipo *QuickDraw*.
- **Avances:**
  - Implementó un prototipo en HTML5/Canvas para capturar dibujos.
  - Integró un demo que muestra **trazos sugeridos en rojo**.
- **Problemas:**
  - El modelo con RNN tiene baja precisión.
  - El sistema muestra todos los trazos demasiado rápido → problema de sincronización.
- **Recomendaciones:**
  - Explorar **modelos en Hugging Face** más adecuados (Transformers para secuencias).
  - Manejar asincronía en Python/JS para controlar la secuencia de trazos.
- **Próximos pasos:** mejorar conexión entre frontend (Canvas) y backend (modelo), refinar la lógica de sincronía.
